```python
from pathlib import Path
import pandas as pd

BASE_DIR = Path(r"E:\Mutual Fund Blue")
DATA_PATH = BASE_DIR / "data" / "processed" / "all_nav_history.csv"

print("Looking for:", DATA_PATH)
print("File exists:", DATA_PATH.exists())

df = pd.read_csv(DATA_PATH)

df["date"] = pd.to_datetime(df["date"])
df["nav"] = pd.to_numeric(df["nav"])

df = df.sort_values(["scheme", "date"])

print("Shape:", df.shape)
print("Schemes:", df["scheme"].unique())

df.head()
```

```python
df["daily_return"] = (
    df.groupby("scheme")["nav"]
      .pct_change()
)

df[["scheme", "date", "nav", "daily_return"]].head(10)
```

```python
df["daily_return"] = (
    df.groupby("scheme")["nav"]
      .pct_change()
)

df[["scheme", "date", "nav", "daily_return"]].head(10)
```

```python
performance = []

for scheme, group in df.groupby("scheme"):

    group = group.sort_values("date").dropna(subset=["daily_return"])

    first_nav = group["nav"].iloc[0]
    last_nav = group["nav"].iloc[-1]
    trading_days = len(group)

    annualized_return = (
        (last_nav / first_nav) ** (252 / trading_days)
    ) - 1

    performance.append({
        "scheme": scheme,
        "first_date": group["date"].iloc[0],
        "last_date": group["date"].iloc[-1],
        "first_nav": first_nav,
        "last_nav": last_nav,
        "trading_days": trading_days,
        "annualized_return": annualized_return
    })

performance_df = pd.DataFrame(performance)

performance_df.round(4)
```

```python
volatility = (
    df.groupby("scheme")["daily_return"]
      .std()
      .mul(np.sqrt(252))
)

performance_df = performance_df.merge(
    volatility.rename("annualized_volatility"),
    on="scheme"
)

performance_df[
    ["scheme", "annualized_return", "annualized_volatility"]
].round(4)
```

```python
volatility = (
    df.groupby("scheme")["daily_return"]
      .std()
      .mul(np.sqrt(252))
)

performance_df = performance_df.merge(
    volatility.rename("annualized_volatility"),
    on="scheme"
)

performance_df[
    ["scheme", "annualized_return", "annualized_volatility"]
].round(4)
```

```python
max_drawdowns = []

for scheme, group in df.groupby("scheme"):
    group = group.sort_values("date").copy()

    running_max = group["nav"].cummax()
    drawdown = (group["nav"] / running_max) - 1

    max_drawdowns.append({
        "scheme": scheme,
        "max_drawdown": drawdown.min()
    })

drawdown_df = pd.DataFrame(max_drawdowns)

performance_df = performance_df.merge(
    drawdown_df,
    on="scheme"
)

performance_df[
    ["scheme", "max_drawdown"]
].round(4)
```

```python
performance_df.round(4)
```

```python
benchmark = "HDFC_Top100_Direct"

returns = df.pivot(
    index="date",
    columns="scheme",
    values="daily_return"
)

benchmark_returns = returns[benchmark]

betas = {}

for scheme in returns.columns:
    if scheme == benchmark:
        betas[scheme] = 1.0
        continue

    combined = pd.concat(
        [returns[scheme], benchmark_returns],
        axis=1
    ).dropna()

    covariance = combined.iloc[:, 0].cov(combined.iloc[:, 1])
    benchmark_variance = combined.iloc[:, 1].var()

    betas[scheme] = covariance / benchmark_variance

beta_df = pd.DataFrame(
    list(betas.items()),
    columns=["scheme", "beta"]
)

performance_df = performance_df.merge(
    beta_df,
    on="scheme"
)

performance_df[["scheme", "beta"]].round(4)
```

```python
var_95 = (
    df.groupby("scheme")["daily_return"]
      .quantile(0.05)
      .rename("VaR_95_daily")
      .reset_index()
)

performance_df = performance_df.merge(
    var_95,
    on="scheme"
)

performance_df[
    ["scheme", "VaR_95_daily"]
].round(4)
```

```python
final_columns = [
    "scheme",
    "first_date",
    "last_date",
    "trading_days",
    "annualized_return",
    "annualized_volatility",
    "sharpe_ratio",
    "max_drawdown",
    "beta",
    "VaR_95_daily"
]

final_metrics = performance_df[final_columns].copy()

final_metrics.round(4)
```

## Performance Metrics Methodology

- Annualized return is calculated using 252 trading days.
- Annualized volatility = daily return standard deviation × √252.
- Sharpe Ratio = annualized return / annualized volatility, using a 0% risk-free rate.
- Beta is calculated relative to HDFC Top 100 Direct.
- Maximum Drawdown measures the largest decline from a previous NAV peak.
- VaR is historical 95% daily VaR, based on the 5th percentile of daily returns.
